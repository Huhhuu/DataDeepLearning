# Tuần 1 - Tải dữ liệu và khám phá dữ liệu

Sử dụng thư viện pandas để đọc dữ liệu từ đường dẫn CSV trực tuyến. Sau đó, chúng ta xem qua 5 dòng đầu tiên, kích thước tổng thể và thông tin kiểu dữ liệu của các cột để có cái nhìn tổng quan ban đầu.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = 'https://raw.githubusercontent.com/huynhhoc/DataAnalystDeepLearning/main/Data/dulieuxettuyendaihoc.csv'
df = pd.read_csv(url)

print("5 dòng đầu tiên của dữ liệu:")
display(df.head())

print(f"\nKích thước dữ liệu: {df.shape}")

print("\nDanh sách cột và kiểu dữ liệu:")
print(df.info())
```

Thực hiện kiểm tra các giá trị bị thiếu (null) trong tập dữ liệu và đếm số lượng dòng bị trùng lặp để đảm bảo tính toàn vẹn của dữ liệu trước khi phân tích.

```python
missing_values = df.isnull().sum()
print("Số lượng giá trị thiếu theo từng cột:")
print(missing_values[missing_values > 0] if missing_values.sum() > 0 else "Không có dữ liệu thiếu")

duplicate_count = df.duplicated().sum()
print(f"\nSố lượng dòng trùng lặp: {duplicate_count}")
```

Sử dụng hàm describe để tính toán các thông số thống kê cơ bản như trung bình, độ lệch chuẩn, giá trị lớn nhất và nhỏ nhất cho các cột dữ liệu số.

```python
print("Thống kê mô tả cho các biến số:")
display(df.describe())
```

Trực quan hóa số lượng giá trị thiếu bằng biểu đồ cột giúp dễ dàng nhận diện những cột nào đang bị thiếu dữ liệu nghiêm trọng.

```python
plt.figure(figsize=(10, 6))
sns.barplot(x=missing_values.index, y=missing_values.values)
plt.xticks(rotation=90)
plt.title('Số lượng giá trị thiếu theo cột')
plt.ylabel('Số lượng')
plt.show()
```
