```python
# @title Xử lý dữ liệu và loại bỏ Outliers bằng Z-Score
import pandas as pd
import numpy as np
import torch

homes = pd.read_csv('https://raw.githubusercontent.com/huynhhoc/DataAnalystDeepLearning/refs/heads/main/Data/23_HOMES.csv')

for col in homes.select_dtypes(include=[np.number]).columns:
    homes[col] = homes[col].fillna(homes[col].median())

target_cols = ['Area', 'List_Price']
outlier_indices = set()

for col in target_cols:
    if col in homes.columns:
        mean = np.mean(homes[col])
        std = np.std(homes[col])

        z_scores = (homes[col] - mean) / std

        mask = np.abs(z_scores) > 3
        outlier_indices.update(homes[mask].index)

homes_clean = homes.drop(list(outlier_indices)).copy()

for col in target_cols:
    data = homes_clean[col]
    mean = np.mean(data)
    std = np.std(data)
    skew = np.mean(((data - mean) / std) ** 3)
    print(f"Skewness của {col}: {skew:.2f}")

features = homes_clean.select_dtypes(include=[np.number]).values
tensor_data = torch.tensor(features, dtype=torch.float32)

print(f"\nKích thước Tensor sau xử lý: {tensor_data.shape}")
print(f"Số dòng bị loại bỏ do Outlier (Z-Score): {len(outlier_indices)}")
```

```python
import pandas as pd
import numpy as np
import torch

homes = pd.read_csv('https://raw.githubusercontent.com/huynhhoc/DataAnalystDeepLearning/refs/heads/main/Data/23_HOMES.csv')

for col in homes.select_dtypes(include=[np.number]).columns:
    homes[col] = homes[col].fillna(homes[col].median())

target_cols = ['Area', 'List_Price']
outlier_indices = set()

for col in target_cols:
    Q1 = homes[col].quantile(0.25)
    Q3 = homes[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    mask = (homes[col] < lower_bound) | (homes[col] > upper_bound)
    outlier_indices.update(homes[mask].index)

homes_clean = homes.drop(list(outlier_indices)).copy()

for col in target_cols:
    data = homes_clean[col]
    mean = np.mean(data)
    std = np.std(data)
    skew = np.mean(((data - mean) / std) ** 3)
    print(f"Skewness của {col}: {skew:.2f}")

features = homes_clean.select_dtypes(include=[np.number]).values
tensor_data = torch.tensor(features, dtype=torch.float32)

print(f"\nKích thước Tensor sau xử lý: {tensor_data.shape}")
print(f"Số dòng bị loại bỏ: {len(outlier_indices)}")
```
