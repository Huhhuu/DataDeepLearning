# Dự Án Phân Tích Dữ Liệu Xét Tuyển Đại Học

Quy trình này thực hiện các bước tiền xử lý dữ liệu để chuẩn bị cho mô hình Deep Learning.



```python
import os
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

# 1. Tạo cấu trúc thư mục lưu trữ
dirs = ['data/raw', 'data/processed']
for d in dirs:
    os.makedirs(d, exist_ok=True)

print("Đã tạo các thư mục:", dirs)
```

### 2. Tải và Khám phá dữ liệu
Chúng ta sẽ tải dữ liệu trực tiếp từ GitHub và lưu bản thô vào thư mục `data/raw`.

```python
url = 'https://raw.githubusercontent.com/huynhhoc/DataAnalystDeepLearning/refs/heads/main/Data/dulieuxettuyendaihoc.csv'
df = pd.read_csv(url)

# Lưu trữ bản thô
df.to_csv('data/raw/raw_data.csv', index=False)

print("Thông tin sơ bộ về dữ liệu:")
display(df.info())
display(df.head())
```

### 3. Làm sạch dữ liệu
Bước này tập trung vào việc xử lý các giá trị rỗng (NaN) và chuyển đổi các cột dữ liệu phù hợp.

```python
import pandas as pd
import numpy as np

# Kiểm tra và xử lý giá trị thiếu
df = df.fillna(df.median(numeric_only=True))

# One-Hot Encoding cho các cột phân loại
categorical_cols = df.select_dtypes(include=['object']).columns
df = pd.get_dummies(df, columns=categorical_cols)

print("Dữ liệu sau khi làm sạch và mã hóa:")
display(df.head())
print(f"Tổng số giá trị thiếu còn lại: {df.isnull().sum().sum()}")
```

### 3.1 Chuyển đổi dữ liệu phân loại (One-Hot Encoding)
Trong Deep Learning, các cột dạng chữ (như Khu vực, Dân tộc) cần được chuyển sang dạng số. One-Hot Encoding sẽ tạo ra các cột 0/1 tương ứng với từng phân loại.

```python
# Xác định các cột không phải dạng số (object/category)
categorical_cols = df.select_dtypes(include=['object']).columns
print(f"Các cột sẽ được mã hóa: {list(categorical_cols)}")

# Thực hiện One-Hot Encoding
df_encoded = pd.get_dummies(df, columns=categorical_cols)

# Cập nhật lại df chính để dùng cho các bước sau
df = df_encoded

print("Kích thước dữ liệu sau khi mã hóa:", df.shape)
display(df.head())
```

### 3.2 Kiểm tra lại giá trị thiếu sau mã hóa
Đảm bảo rằng quá trình xử lý không để lại giá trị rỗng nào.

```python
final_missing_check = df.isnull().sum().sum()
print(f"Tổng số giá trị thiếu còn lại: {final_missing_check}")

if final_missing_check > 0:
    print("Các cột vẫn còn giá trị thiếu:")
    display(df.columns[df.isnull().any()].tolist())
else:
    print("Dữ liệu đã sạch hoàn toàn.")
```

### 4. Chia tập Train và Test
Thông thường chúng ta chia theo tỷ lệ 80% Train và 20% Test.

```python
# Chia dữ liệu
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)

# Lưu vào thư mục processed
train_df.to_csv('data/processed/train.csv', index=False)
test_df.to_csv('data/processed/test.csv', index=False)

print(f"Kích thước tập Train: {train_df.shape}")
print(f"Kích thước tập Test: {test_df.shape}")
print("Dữ liệu đã sẵn sàng trong thư mục 'data/processed/'")
```
