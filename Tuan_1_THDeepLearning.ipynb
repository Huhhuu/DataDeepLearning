### Phân loại dữ liệu và Xác định thang đo
- **Dữ liệu định tính (Qualitative):** `GT` (Nominal), `KV` (Nominal), `KT` (Nominal), `DT` (Nominal).
- **Dữ liệu định lượng (Quantitative):** `T1..T6`, `L1..L6`, `H1..H6`, `S1..S6`, `V1..V6`, `X1..X6`, `D1..D6`, `N1..N6` (Ratio), `DH1..DH3` (Ratio).

```python
import pandas as pd
import numpy as np

url = 'https://raw.githubusercontent.com/huynhhoc/DataAnalystDeepLearning/refs/heads/main/Data/dulieuxettuyendaihoc.csv'
df = pd.read_csv(url)

print("10 dòng đầu tiên:")
display(df.head(10))
print("\n10 dòng cuối cùng:")
display(df.tail(10))
```

### Xử lý dữ liệu thiếu
- Cột `DT` (Dân tộc): Điền 0.
- Các cột điểm số: Điền giá trị trung bình (Mean).

```python
print("Tần số giá trị unique cột DT trước xử lý:", df['DT'].unique())
df['DT'] = df['DT'].fillna(0)

print(f"Số lượng missing T1: {df['T1'].isnull().sum()}")
df['T1'] = df['T1'].fillna(df['T1'].mean())

score_cols = ['T1','L1','H1','S1','V1','X1','D1','N1','T2','L2','H2','S2','V2','X2','D2','N2','T3','L3','H3','S3','V3','X3','D3','N3','T4','L4','H4','S4','V4','X4','D4','N4','T5','L5','H5','S5','V5','X5','D5','N5','T6','L6','H6','S6','V6','X6','D6','N6','DH1','DH2','DH3']
for col in score_cols:
    df[col] = df[col].fillna(df[col].mean())

print("Tổng missing value sau xử lý:", df.isnull().sum().sum())
```

### Tính toán TBM, Xếp loại, và Quy đổi thang điểm Mỹ

```python
for i in range(1, 4):
    df[f'TBM{i}'] = (df[f'T{i}']*2 + df[f'L{i}'] + df[f'H{i}'] + df[f'S{i}'] + df[f'V{i}']*2 + df[f'X{i}'] + df[f'D{i}'] + df[f'N{i}']) / 10

def xep_loai(diem):
    if diem < 5.0: return 'Y'
    elif diem < 6.5: return 'TB'
    elif diem < 8.0: return 'K'
    elif diem < 9.0: return 'G'
    else: return 'XS'

for i in range(1, 4):
    df[f'XL{i}'] = df[f'TBM{i}'].apply(xep_loai)

for i in range(1, 4):
    min_val, max_val = df[f'TBM{i}'].min(), df[f'TBM{i}'].max()
    df[f'US_TBM{i}'] = ((df[f'TBM{i}'] - min_val) / (max_val - min_val)) * 4

display(df[['TBM1', 'XL1', 'US_TBM1']].head())
```

### Kết quả xét tuyển (KQXT)
Tính toán dựa trên tổ hợp môn của từng khối thi.

```python
def tinh_kqxt(row):
    kt = row['KT']
    dh1, dh2, dh3 = row['DH1'], row['DH2'], row['DH3']

    if kt in ['A', 'A1']:
        diem = (dh1*2 + dh2 + dh3) / 4
    elif kt == 'B':
        diem = (dh1 + dh2*2 + dh3) / 4
    else:
        diem = (dh1 + dh2 + dh3) / 3

    return 1 if diem >= 5.0 else 0

df['KQXT'] = df.apply(tinh_kqxt, axis=1)
print("Số lượng thí sinh Đậu (1) và Rớt (0):")
print(df['KQXT'].value_counts())
```

```python
df.to_csv('processed_dulieuxettuyendaihoc.csv', index=False)
print("Đã lưu file: processed_dulieuxettuyendaihoc.csv")
```

### 3. Làm sạch dữ liệu
Bước này tập trung vào việc xử lý các giá trị rỗng (NaN) và chuyển đổi các cột dữ liệu phù hợp.

```python
import pandas as pd
import numpy as np

categorical_cols = df.select_dtypes(include=['object']).columns
df = pd.get_dummies(df, columns=categorical_cols)

for col in df.select_dtypes(include=['bool']).columns:
    df[col] = df[col].astype(int)

print("Dữ liệu sau khi làm sạch và mã hóa:")
display(df.head())
print(f"Tổng số giá trị thiếu còn lại: {df.isnull().sum().sum()}")
```

### 3.1 Kiểm tra lại giá trị thiếu sau mã hóa
Đảm bảo rằng quá trình xử lý không để lại giá trị rỗng nào.

```python
final_missing_check = df.isnull().sum().sum()
print(f"Tổng số giá trị thiếu còn lại: {final_missing_check}")

if final_missing_check > 0:
    print("Các cột vẫn còn giá trị thiếu:")
    display(df.columns[df.isnull().any()].tolist())
else:
    print("Dữ liệu đã sạch hoàn toàn.")
```

### 4. Chia tập Train và Test
Thông thường chúng ta chia theo tỷ lệ 80% Train và 20% Test.

```python
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)

train_df.to_csv('data/processed/train.csv', index=False)
test_df.to_csv('data/processed/test.csv', index=False)

print(f"Kích thước tập Train: {train_df.shape}")
print(f"Kích thước tập Test: {test_df.shape}")
print("Dữ liệu đã sẵn sàng trong thư mục 'data/processed/'")
```
